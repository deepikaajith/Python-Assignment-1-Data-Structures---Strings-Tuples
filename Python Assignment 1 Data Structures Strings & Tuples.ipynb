```python
# Q1. Concatenate "Hello " with the user's name.

'''This program joins two or more pieces of text together. First, I store "Hello " in one variable and "Zara" in another variable. 
The + symbol is used to join them, and the space after "Hello " makes sure there is a gap between Hello and Zara. 
Then, another message "welcome to Python programming" is added to display the complete sentence'''

t1 = "Hello "
t2 = input("Enter your Name ", )
print(t1 + t2)

# Q2. Add ", welcome to Python programming" to the existing string.

t3 = (" Welcome to Python Programming")
t = t1+t2+t3
print(t)
```

```python
# a.Print the first character of the string?
# Since the index starts from 0, I used index 0 to print the first character.

print(t[0])

#b. Print the last character of the string?
# as index -1 stands for last character, I used index -1 to print it.

print(t[-1])

#c. Print the first 5 characters of the string?
# Print the first 5 characters using slicing.

print(t[0:5])

#d.Print the last 11 characters of the string?
#here -11  means start 11 characters from the end.

print(t[-11:])

#e. Print the string in reverse?
#[::-1] reverses the string from last character to first.

print(t[: : -1])

#f. Use slicing and print the word “Python” from the existing string.
# The index of 'P' is 22, so I used 28 as the ending index to print "Python".

print(t[22:28])
```

```python
#Write a Python program that takes a string, strM = “Python beginner tutorial” and perform the following tasks:
#a. Convert the sentence to uppercase.

strM = "Python beginner tutorial"
print(strM.upper())

#b. Convert the sentence to lowercase.
print(strM.lower())

#c. Use Capitalize and return the sentence to the original input form.

print(strM.capitalize())

#d. Count the total number of occurrences of character ‘t’ in the string.

print(strM.count("t"))

#e. Replace all occurrences of “Python” with “Machine Learning” in the input string strM = “Python beginner tutorial”

strM = strM.replace("Python" , "Machine Learning")
print(strM)
```
